Agora que temos informações sobre os dias do mês e da semana, vamos ver se é possível encontrar alguma relação entre os resultados encontrados.

```python
import calendar
from datetime import date

# 1. Configurar o ano de análise
ANO = 2024

# 2. Mapeamento de nomes de dias (de 0=Segunda a 6=Domingo)
dias_pt = {
    0: 'Segunda-feira', 1: 'Terça-feira', 2: 'Quarta-feira',
    3: 'Quinta-feira', 4: 'Sexta-feira', 5: 'Sábado', 6: 'Domingo'
}

# 3. Inicializar a estrutura de dados (Tabela de Contagem)
# As chaves são os dias do mês (1 a 31). Os valores serão dicionários de contagem de dias.
contagem_dias = {dia_mes: {nome_dia: 0 for nome_dia in dias_pt.values()} for dia_mes in range(1, 32)}

# 4. Iterar sobre todos os meses do ano de 2024
for mes in range(1, 13):
    # calendar.Calendar(0) começa a semana na Segunda-feira (0)
    cal = calendar.Calendar(0)
    
    # Gera uma lista de tuplas (dia_do_mes, dia_da_semana_indice)
    mes_info = cal.itermonthdays2(ANO, mes)
    
    for dia_mes, dia_semana_indice in mes_info:
        # 0 significa que o dia_mes não existe naquele mês (é preenchimento do calendário), ignorar
        if dia_mes == 0:
            continue
        
        # O dia 29, 30 e 31 não existem em todos os meses
        if dia_mes in contagem_dias:
            nome_dia_semana = dias_pt[dia_semana_indice]
            contagem_dias[dia_mes][nome_dia_semana] += 1

# 5. Converter o resultado para um DataFrame Pandas
# Transforma o dicionário aninhado em um DataFrame
df_calendario = pd.DataFrame(contagem_dias).T 

# 6. Limpar e formatar o DataFrame
# Adicionar o Dia do Mês como coluna
df_calendario.index.name = 'Dia_do_Mês'

# Preencher NaN (dias 29, 30, 31 em meses curtos) com 0 e converter para inteiro
df_calendario = df_calendario.fillna(0).astype(int)

# 7. Exibir o resultado
print(f"--- Frequência de Dias da Semana por Dia do Mês (Ano {ANO}) ---")
df_calendario
```

Analisando a tabela acima em conjunto com o gráfico de faturamento médio por dia, não podemos tirar análises muito conclusivas. Vemos, por exemplo, que os dias 14 e 21 (dias de alta na rua e baixa no shopping) estão entre os dias que coincidem com o fim de semana (dias de faturamento maior no shopping) com mais frequência. De qualquer forma, vemos **comportamentos bem variados ao longo do mês e da semana**, que devem ser levados em conta no modelo.

Diante das análises acima, o **principal achado** é que os **meses de marços, novembro e dezembro** podem atuar como **possíveis outliers**. Sendo assim, a partir de uma análise aprofundada dos produtos, queremos entender se essa observação é confirmada.

```python
df_preco_categoria = df_agg.groupby('CATEGORIA').agg(
    Faturamento_Total=('FATUR_VENDA', 'sum'),
    Qtd_Total=('QTD_VENDA', 'sum')
).reset_index()

df_preco_categoria['Preco_Medio_Categoria'] = df_preco_categoria['Faturamento_Total'] / df_preco_categoria['Qtd_Total']
df_preco_categoria.sort_values(by='Preco_Medio_Categoria', ascending=False, inplace=True)

df_tabela_final = df_preco_categoria[
    ['CATEGORIA', 'Preco_Medio_Categoria', 'Faturamento_Total', 'Qtd_Total']
]

print(df_tabela_final.to_string(index=False))
```

Algo a se notar é que Lanchonete, apesar de ser a terceira categoria com maior volume de vendas, o faturamento total é baixo, levando a um preço médio baixo. Ao longo das nossas análises, notamos algo que poderia explicar essa diferença de volume.

```python
# Divisão por unidade de medida
df_un = df_agg[df_agg['UNID'] == 'UN'].drop_duplicates(subset='SKU', keep='first').reset_index(drop=True)
df_un = df_un[['NOME_PRODUTO', 'CATEGORIA']]

df_kg = df_agg[df_agg['UNID'] == 'KG'].drop_duplicates(subset='SKU', keep='first').reset_index(drop=True)
df_kg = df_kg[['NOME_PRODUTO', 'CATEGORIA']]

# Visualizar resultados
print(f"Linhas únicas (UNID=='UN'): {len(df_un)}")
print(f"Linhas únicas (UNID=='KG'): {len(df_kg)}\n")

# Análise quantitativa por unidade de medida
df_unid_analise = df_agg.groupby('UNID').agg(
    Faturamento_Total=('FATUR_VENDA', 'sum'),
    Qtd_Total=('QTD_VENDA', 'sum'),
    Num_Transacoes=('COD_VENDA', 'nunique')
).reset_index()

df_unid_analise['Qtd_Media_Item'] = df_unid_analise['Qtd_Total'] / df_unid_analise['Num_Transacoes']
print(df_unid_analise[['UNID','Qtd_Total', 'Num_Transacoes', 'Qtd_Media_Item']].to_string(index=False))

print("\nProdutos com UNID 'UN':")
df_un.head()
```

Implicações de Negócio
